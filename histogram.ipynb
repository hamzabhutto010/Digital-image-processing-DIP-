```python
# histogram of an image
img = cv2.imread('images/fig2.tif', 0)
img_flaten = img.flatten()
freq = np.zeros((256))

for i in img_flaten:
    freq[i] = freq[i] + 1

unique_pixel = np.arange(256)

plt.bar(unique_pixel, freq)
plt.xlabel('intensites')
plt.ylabel('frequency')
plt.title('image histogram')
```

```python
# histogram equalization
img = cv2.imread('images/fig2.tif', 0)

freq = np.zeros((256))
flt_img = img.flatten()

# histogram of image
for i in flt_img:
    freq[i] = freq[i] + 1

# histogram normalization

pdf = freq / len(flt_img)

# comulative distribution function
cdf = np.zeros_like(pdf)

for i in range(len(pdf)):
    cdf[i] = sum(pdf[:i + 1])

# to map old pixel replace by new pixel value
equ_img = np.zeros_like(img)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        equ_img[i,j] = cdf[img[i,j]] * 255

# histogram equalized
fig = plt.figure(figsize=(20,5))

plt.subplot(1,2,1)
plt.bar(np.arange(256),freq)
plt.xlabel('intensites')
plt.ylabel('frequency')
plt.title('histogram of original image')

plt.subplot(1,2,2)
plt.bar(np.round(cdf*255),freq)
plt.xlabel('intensities')
plt.ylabel('frequency')
plt.title("histogram equalized")
    
```

```python
# historgram equalization 
img = cv2.imread('images/moon.jpeg')

flat_img = img.flatten()
freq = np.zeros((256))

for i in flat_img:
    freq[i] = freq[i] + 1
    
# normalize the histogram
pdf = freq / len(flat_img)

# cdf calculation
cdf = np.zeros_like(pdf)

for i in range(len(pdf)):
    cdf[i] = sum(pdf[:i + 1])

#map old pixel with new one
equ_img = np.zeros_like(img)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        equ_img[i,j] = cdf[img[i,j]] * 255

fig = plt.Figure(figsize=(8,10))

plt.subplot(2,2,1)
plt.imshow(img)
plt.axis('off')
plt.title('original image')

plt.subplot(2,2,2)
plt.imshow(equ_img)
plt.title('equalized image')
plt.axis('off')

fig1 = plt.figure(figsize=(20,5))

plt.subplot(1,2,1)
plt.bar(np.arange(256),freq)
plt.xlabel('intensites')
plt.ylabel('frequency')
plt.title('histogram of original image')

plt.subplot(1,2,2)
plt.bar(np.round(cdf*255),freq)
plt.xlabel('intensities')
plt.ylabel('frequency')
plt.title("histogram equalized")

# 
     
```
